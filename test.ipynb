```python
import numpy as np
from Util.statistics_util import *

n = np.array([0,1,2])
g = np.array([3,400,5])

stats = uniform_statistics(ground_truth_measurements=g, new_system_measurements=n)

for key, value in stats.items():
    print(key, value)
```

```python
mean_g = np.mean(g)
mean_n = np.mean(n)
# absolute error
absolute_error = np.mean(np.abs(g - n))
# relative error
relative_error = (
    np.mean(np.abs((g - n) / g)) * 100
)
# Calculate RMSE
rmse = np.mean(np.sqrt(np.mean((g - n) ** 2)))
# relative RMSE
relative_rmse = rmse / np.mean(g)

result = f'mean gt: {mean_g}, mean ns: {mean_n}\nabs error: {absolute_error}, rel error: {relative_error} \nabs rmse: {rmse}, rel rmse: {relative_rmse}'
print(result)
```
